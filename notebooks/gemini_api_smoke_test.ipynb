# Gemini API key and model check

This notebook makes one small Gemini request using `GEMINI_API_KEY` and `GEMINI_MODEL` from the project `.env` (or the process environment). It never prints the key. The prompt and evidence below are synthetic.

Run it from the project virtual environment. If Google returns `503 UNAVAILABLE`, the credentials were sent but the model service may be temporarily overloaded; retry later.

```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
GEMINI_API_KEY = os.getenv("GEMINI_API_KEY")
if not GEMINI_API_KEY:
    raise RuntimeError("Set GEMINI_API_KEY in the project .env file or process environment.")

model_name = os.getenv("GEMINI_MODEL", "gemini-3.6-flash")
llm = ChatGoogleGenerativeAI(model=model_name, api_key=GEMINI_API_KEY)

response = llm.invoke("Reply with exactly: Gemini API connection successful")
print(f"Model: {model_name}")
print(response.content)
```

```python
response = llm.invoke("What is the capital of India")

response
```
